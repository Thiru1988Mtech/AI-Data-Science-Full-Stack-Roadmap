## List

```python
cities = ['chennai', 'bangalore', 'hyderabad']
print(type(cities))
```

```python
temperatures = [30, 20, 35]
print(type(temperatures))
```

```python
print('No of values in cities', len(cities))
print('No of values in temperatures %d' % len(temperatures))
```

```python
print(cities[0]) # Access first element inside the list
print(cities[2]) # Access third element inside the list
print(cities[len(cities)-1]) # Access last element of the list
```

```python
print(cities[2])
print(cities[len(cities)-1])
print(cities[-1]) # Negative Indexing
```

```python
# Access first two values
ages = [10, 30, 50, 70]
print(ages[0:3]) # Slice first three elements
print(ages[:3]) # Slice first three elements. Shortcut
```

```python
ages = [10, 50, 40, 60, 55, 33]
ages[2:5]
```

```python
print(ages[-3:])
```

```python
cities = ['chennai', 'bangalore', 'hyderabad']
cities[0] = 'mumbai' # Change a particular value based on index
cities
```

```python
cities = ['chennai', 'bangalore', 'hyderabad']
del cities[0] # Removing the element based on its index
cities
```

### Loops
- For loop
- While loop

```python
## For loop to access every element in a list
cities = [10, 40, 50, 43]
for i in range(len(cities)):
  print(i, cities[i])
print('-----')
```

```python
for index in range(len(cities)):
  print(index, cities[index])
  print('-----')
```

```python
cities = ['chennai', 'bangalore', 'hyderabad', 'chennai']
for index in range(len(cities)):
  print(index, cities[index])
```

```python
cities = ['chennai', 'bangalore', 'hyderabad', 'chennai']
for city in cities:
  print(city, len(city))
```

```python
## Finding minimum value in a list manually
## Best way: min(list_name)
ages = [10, 40, 45, 5, 30]
min(ages)
```

```python
## Not optimal way
min_val = ages[0]
for age in ages:
  if age < min_val:
    min_val = age
print(min_val)
```
