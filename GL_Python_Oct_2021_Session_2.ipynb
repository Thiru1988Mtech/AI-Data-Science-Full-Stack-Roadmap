```python
# Exercise: Calculate number of centuries for each player
players = {'sachin': [50, 30, 100, 200, 20],
           'rahul': [40, 50, 30, 100],
           'dhoni': [60, 30, 21]}
# Expected output: {'sachin': 2, 'rahul': 1, 'dhoni': 0}
result = {}
for player_name_str, runs_list in players.items():
  counter = 0
  for runs_int in runs_list:
    if runs_int >= 100:
      counter = counter + 1
  result.update({player_name_str: counter})
result
```
