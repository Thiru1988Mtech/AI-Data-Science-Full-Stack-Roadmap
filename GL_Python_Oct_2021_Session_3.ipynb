### Regular Functions

```python
def square_list(x):
  y = [pow(i, 2) for i in x ]
  return y

z = [1, 2, 3, 4]
result = square_list(z)
result
```

```python
# Exercise:
# Create a funtion as `compute_hashtag_freq`
# which will take list of tweets (string) as input
# Returns a series which contains frequency of each hashtag
```

```python
def compute_hashtag_freq(docs):
  all_hashtags = []
  for tweet in docs:
    words_clean = re.findall('[a-zA-Z0-9#]+', tweet.lower())
    hashtags = [word for word in words_clean if word.startswith('#') and word != '#']
    all_hashtags.extend(hashtags)
  hashtags_freq = pd.Series(all_hashtags).value_counts()
  return hashtags_freq
result = compute_hashtag_freq(tweets['text'])
result.head()
```

### Lamdba functions

```python
# regular function
def square_list(x):
  y = [pow(i, 2) for i in x ]
  return y
res = square_list([10, 20])
res
```

```python
# lambda function
square_list = lambda x: [pow(i, 2) for i in x ]
res = square_list([10, 20])
res
```

```python
compute_hashtag_freq = lambda docs: pd.Series([word for tweet in docs for word in re.findall('[a-zA-Z0-9#]+', tweet.lower()) if word.startswith('#') and word != '#']).value_counts()
result = compute_hashtag_freq(tweets['text'])
result.head()
```
